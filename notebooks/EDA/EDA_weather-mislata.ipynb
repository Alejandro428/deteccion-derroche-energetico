```python
query = f"""
SELECT
    time AS timestamp,
    entity_id,
    state,
    attributes
FROM {tabla}
"""

df = pd.read_sql(query, engine)

print(df.head())
print(df.shape)
```

# Weather

```python
sensores_mislata=df["entity_id"].str.contains("mislata")
df_mislata = df[sensores_mislata].copy()
df_mislata.entity_id.unique()
```

```python
#df_mislata, los quiero con pivot table por entity_id
df_mislata["state"] = pd.to_numeric(df_mislata["state"], errors="coerce")
df_mislata_wide=df_mislata.pivot(index="timestamp", columns="entity_id", values="state")
df_mislata_wide = df_mislata_wide.resample("1H").mean()
df_mislata_wide = df_mislata_wide.reset_index()

df_mislata_wide["date"] = df_mislata_wide["timestamp"].dt.date
df_mislata_wide["hour"] = df_mislata_wide["timestamp"].dt.hour
df_mislata_wide=df_mislata_wide.set_index('timestamp')
df_mislata_wide
```

```python
#quitamos weather.forecast_mislata
df_mislata_wide = df_mislata_wide.drop(columns=["weather.forecast_mislata"])
df_mislata_wide = df_mislata_wide.rename(columns={
    "sensor.mislata_temperatura": "temperatura",
    "sensor.mislata_humedad": "humedad",
    "sensor.mislata_presion": "presion",
    "sensor.mislata_viento": "viento",
    "sensor.mislata_direccion_viento": "direccion_viento",
    "sensor.mislata_nubosidad": "nubosidad"
})
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# columnas de sensores (excluimos columnas auxiliares)
sensor_cols = df_mislata_wide.columns.difference(["date","hour"])

# matriz de disponibilidad (1 = hay dato, 0 = NaN)
availability = df_mislata_wide[sensor_cols].notna().astype(int)

# transponer para tener sensores en filas
availability = availability.T

plt.figure(figsize=(16,6))

sns.heatmap(
    availability,
    cmap=["red","blue"],
    cbar=False
)

plt.title("Disponibilidad de datos por sensor (verde = dato presente)")
plt.xlabel("Tiempo")
plt.ylabel("Sensor")

plt.tight_layout()
plt.show()
```

En promedio los huecos son de 9h, en promedio los huecos son muy pequeÃ±os y el 75% duran menos de 1H y hay un hueco muy grande. Asi que lo suyo es eliminar los huecos de 24h y el resto los interpolamos

Como tenemos un 18% 20% de datos faltantes por sensor no podemos eliminarlos, enonces una opcion es interpolar los datos que faltan ya que los datos que faltan no son tramos muy extensos

```python
sensor_cols = df_mislata_wide.columns.difference(["date","hour"])

# detectar filas completamente vacÃ­as
rows_all_nan = df_mislata_wide[sensor_cols].isna().all(axis=1)

# eliminar huecos gigantes (>24h seguidas)
gap_groups = (rows_all_nan != rows_all_nan.shift()).cumsum()
gap_sizes = rows_all_nan.groupby(gap_groups).transform("sum")

df_mislata_wide = df_mislata_wide[~(rows_all_nan & (gap_sizes > 24))]

# interpolar huecos pequeÃ±os
df_mislata_wide[sensor_cols] = df_mislata_wide[sensor_cols].interpolate(
    method="time",
    limit=3
)
```
