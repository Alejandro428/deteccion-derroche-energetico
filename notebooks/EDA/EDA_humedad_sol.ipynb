```python
hum_virtual = df_hum[
    df_hum["entity_id"] == "sensor.sensor_temperatura_2_humidity"
].copy()

hum_virtual["timestamp"] = pd.to_datetime(hum_virtual["timestamp"])
hum_virtual["state"] = pd.to_numeric(hum_virtual["state"], errors="coerce")

hum_virtual = hum_virtual.set_index("timestamp")

# quedarte solo con la columna numÃ©rica
hum_virtual = hum_virtual[["state"]]

hum_virtual = hum_virtual.resample("1H").mean()
```

```python
# eliminar valores imposibles
hum_virtual = hum_virtual[
    (hum_virtual['state'] >= 0) &
    (hum_virtual['state'] <= 100)
]

# ordenar
hum_virtual = hum_virtual.sort_index()

# --- 2. quedarnos solo con humedad ---
df_hum_virtual = hum_virtual[["state"]].copy()

# eliminar nulos
df_hum_virtual = df_hum_virtual.dropna()

# --- 3. construir intervalos de validez ---
df_hum_virtual['next_time'] = df_hum_virtual.index.to_series().shift(-1)
df_hum_virtual
```

```python

# Ãºltimo registro dura hasta el final de la Ãºltima hora
end_time = df_hum_virtual.index.max().ceil('H')
df_hum_virtual.loc[df_hum_virtual['next_time'].isna(), 'next_time'] = end_time
df_hum_virtual
```

```python
# --- 4. funciÃ³n para media ponderada por hora ---
def hourly_weighted_mean(df):

    start = df.index.min().floor('H')
    end = df['next_time'].max().ceil('H')

    hours = pd.date_range(start, end, freq='H')

    results = []

    for h_start in hours[:-1]:
        h_end = h_start + pd.Timedelta(hours=1)

        # intervalos que intersectan con la hora
        mask = (df.index < h_end) & (df['next_time'] > h_start)
        subset = df.loc[mask]

        if subset.empty:
            results.append(np.nan)
            continue

        total_weight = 0
        weighted_sum = 0

        for t, row in subset.iterrows():

            start_i = max(t, h_start)
            end_i = min(row['next_time'], h_end)

            seconds = (end_i - start_i).total_seconds()

            weighted_sum += row['state'] * seconds
            total_weight += seconds

        results.append(weighted_sum / total_weight)

    return pd.DataFrame(
        {'state': results},
        index=hours[:-1]
    )

# --- 5. calcular temperatura horaria ---
hum_hourly = hourly_weighted_mean(df_hum_virtual)

# --- resultado ---
hum_hourly
```
